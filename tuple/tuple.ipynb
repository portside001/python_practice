# Tuple Creation and its theory

- Tuple is an ordered, immutable collection in Python. 
- It supports indexing, slicing, duplicates, and iteration.
  Because tuples are immutable, 
- they are more memory efficient and slightly faster than lists,
-  making them suitable for fixed data 
- such as coordinates, database records, and dictionary keys
- but you can not update 

```python
t = ()
type(t)
```

```python
# this is an exception of tuple python think its int 
# to avoid thi we need to use (10,)

t = (10)
type(t)
```

```python
t = (10,)
type(t)
```

# Tuple = Read Only

Can Do:
✅ Read
✅ Search
✅ Iterate
✅ Count
✅ Index

Cannot Do:
❌ Update
❌ Insert
❌ Delete
❌ Sort
❌ Reverse
❌ Append
❌ Pop

```python
# tuple of integer number
tup = (10,20,30)
print(type(tup),'<===>',tup)
```

```python
# tuple of float number
tup = (10.11,11.12,12.13,12.14)
print(type(tup),'<===>',tup)
```

```python
# tuple of string
tup = ("one","two","three")
```

```python
# nested tuple
tup = ('jhon',25,('michele',30),('don',32))
print(type(tup),'<===>',tup)
```

```python
# tuple of mixed data type
tup = ('jhon',25,('michel',30),{'name','david'},(99,22,33))
print(type(tup))
```

```python
len(tup)
```

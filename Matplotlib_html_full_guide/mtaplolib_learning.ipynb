# Q. What are the APIs of Matplotlib?

Answer:

Matplotlib mainly provides two APIs:

- 1) Pyplot Interface (State-based) -: Easy to use, MATLAB-style interface where plt maintains the current figure and axes automatically. Best for quick plotting and beginners.
- 2) Object-Oriented Interface –:  More powerful and flexible. We explicitly work with Figure and Axes objects (fig, ax). It is recommended for complex visualizations and production code.

There was also a Pylab Interface, which combined Pyplot and NumPy, but it is now discouraged because it imports many functions into the global namespace, making code harder to maintain.

# line garph

```python
months = ['Jan','Feb','Mar','Apr','May']
sales = [10,20,15,30,25]

plt.plot(
    months,
    sales,
    color='red',
    linewidth=3,
    linestyle='--',
    marker='o',
    markersize=8,
    label='Sales'
)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.grid(True)
plt.legend()

plt.show()
```

# bar graph 

```python
months = ['Jan','Feb','Mar']

sales = [100,120,80]
profit = [30,40,20]

x = np.arange(len(months))

plt.bar(x-0.2, sales, width=0.4, label='Sales')
plt.bar(x+0.2, profit, width=0.4, label='Profit')

plt.xticks(x, months)
plt.legend()
plt.show()
```

# scatter plot

```python
study=[1,2,3,4,5]

marks=[30,40,60,70,90]

plt.scatter(
    study,
    marks,
    color='red',
    s=150,
    marker='o',
    alpha=0.8
)

plt.title("Study vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.grid(True)

plt.show()
```

# Histogram

```python
marks=[12,15,18,20,22,25,28,30,35,40,42,45,50]

plt.hist(
    marks,
    bins=5,
    color='skyblue',
    edgecolor='black'
)

plt.title("Student Marks")

plt.xlabel("Marks")

plt.ylabel("Number of Students")

plt.show()
```

# pie chart

```python
expense=[4000,12000,3000,1000]

labels=['Food','Rent','Shopping','Travel']

explode=(0,0.1,0,0)

plt.pie(
    expense,
    labels=labels,
    autopct='%1.1f%%',
    explode=explode,
    shadow=True,
    startangle=90
)

plt.title("Monthly Expenses")

plt.show()
```

# box plot 

```python
salary = [25000,27000,28000,30000,32000,34000,35000,500000]

plt.boxplot(salary,vert=False,showmeans=True,widths=0.4)

plt.title("Employee Salary")

plt.show()
```

```python
salary = [25000, 27000, 28000, 30000, 32000, 34000, 35000, 500000]

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# 1. Full Dataset (shows outlier clearly)
axes[0].boxplot(salary, vert=False, showmeans=True, widths=0.4)
axes[0].set_title("Full Dataset (With Executive Outlier)")
axes[0].set_xlabel("Salary ($)")

# 2. Filtered Dataset (zoomed to regular salaries under 100,000)
regular_salaries = [s for s in salary if s < 100000]
axes[1].boxplot(regular_salaries, vert=False, showmeans=True, widths=0.4)
axes[1].set_title("Zoomed View: Regular Staff ($25k - $35k)")
axes[1].set_xlabel("Salary ($)")

plt.tight_layout()
plt.show()
```

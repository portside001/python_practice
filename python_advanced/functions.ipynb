- using def we create a function is known as user define functions

```python
def greet():
    print('hello function')

greet()
```

```python
def greet(name):
    print(f"Hello {name}")  # Only prints text, does not return a value

result = greet("Alice")
print(result)  # Outputs: None
```

```python
type(result)
```

-- function have formal Arguments and actual arguments 

```python
def add(x,y):   # formal arguments 
    c = x + y
    print(c)
r = add(2,3)  # actual arguments 
```

# Keyword arguments 

```python
def person(name , age, mob):
    print(name)
    print(age)
    print(mob)

person(age = 23 ,name = 'nit',12345)
```

```python
def person(name , age, mob):
    print(name)
    print(age)
    print(mob)

person(age=31,name='nit',mob=12345)
```

# Default arguments 

```python
def person(name, age=18, mob=+91): # default parameters
    print(name)
    print(age)
    print(mob)

person('text')
```

# variable length

```python
def sum(a,b):
    c = a + b
    print(c)
sum(1,2)
```

```python
def sum(a,b):
    c = a + b
    print(c)
sum(1,2,3,4,5,6,6)
```

```python
def sum(a,*b):    # positional arguments take as a tuple here when we add * args 
    c = a + b
    print(c)
    
sum(1,2,3,4,5,6,6)
```

```python
def sum(a,*b):    # positional arguments take as a tuple here when we add * args 
    c = a
    for i in b:
        c = c + i
    print(c)
sum(1,2,3,4,5,6,6)
```

# recurssion 

- function is calling function itself

```python
def factorial(n):
    if n <= 1:
        return 1
    else:
        return n * factorial(n-1)

factorial(5)
```

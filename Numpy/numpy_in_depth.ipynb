# what is numpy 
- numpy is python library that provides mutidimensional array object
- meaning
- Scientific Computing in python 

- "The core object of NumPy is ndarray, which stands for n-dimensional array. It stores data in one or more dimensions and generally requires homogeneous data types. Because the data is stored contiguously in memory, NumPy operations are much faster and more memory efficient than Python lists."

# Numpy Arrays Vs Python Sequences
NumPy arrays have a fixed size at creation, unlike Python lists (which can grow dynamically).
Changing the size of an ndarray will create a new array and delete the original.

Used Everywhere in Data Science

Libraries internally convert data into NumPy arrays.

Examples:

Pandas
Scikit-Learn
OpenCV
TensorFlow
Matplotlib

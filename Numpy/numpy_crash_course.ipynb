```python
num_array = np.arange(100).reshape(20, 5)
print(num_array)

# Display array in Fortran order 

print("\n  C type array:")
for num_array in np.nditer(num_array, order="C"):
	print(num_array,end=' ')
```

```python
# creating a Numpy array
num_array = np.arange(100).reshape(20, 5)
print(num_array)

# Display array in Fortran order 

print("\n  F type array:")
for num_array in np.nditer(num_array, order="F"):  # coulumn wise print
	print(num_array,end=' ')
```

```python
# python programe to generate opt
import random
def generate_otp(length=4):
    """Generate a numeric OTP of a specified length"""
    digit ='012345'
    otp = ''.join(random.choice(digit) for _ in range(length))
    return otp
otp_length = 4 
generate_otp(otp_length)
```

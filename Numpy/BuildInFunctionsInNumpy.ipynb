# 4) Statistical Functions

```python
# mean using np
s = np.array([1,2,3,4])
mean = np.mean(s)
mean
```

```python
# variance of an array
var =np.var(s)
var
```

```python
# standard deviation of an array 
std_dev = np.std(s)
std_dev
```

- Mean      = Sum of values / Number of values

- Variance  = Average of squared deviations from mean

- Std Dev   = √Variance

- Median    = Middle value after sorting

```python
median = np.median(s)
median
```

- A.T                      # transpose
- A @ B                    # matrix multiplication
- np.linalg.det(A)         # determinant
- np.linalg.inv(A)         # inverse
- np.linalg.solve(A,B)     # solve AX=B
- np.linalg.eig(A)         # eigen values/vectors
- np.linalg.matrix_rank(A) # rank
- np.linalg.norm(A)        # norm
- np.linalg.svd(A)         # SVD

- det     → Matrix invertible or not
- inv     → Matrix inverse
- eig     → PCA, ML
- solve   → AX = B
- norm    → Distance/Magnitude
- svd     → Dimensionality reduction
- rank    → Number of independent rows/columns

# 7) Boolean & Logical Functions

```python
logical_arr = np.array([True,False,True])
all_true = np.all(logical_arr)  # check all elements are true 
print("all elements are true\n",all_true)
```

```python
logical_arr = np.array([True,True,True])
all_true = np.all(logical_arr) # check all elements are true 
print("all elements are true\n",all_true)
```

```python
# check if any elemets are true
logical_arr = np.array([True,True,True])
all_true = np.any(logical_arr) # check any elemets are true 
print("all elements are true\n",all_true)
```

# 8) Set Operations

```python
set_a = np.array([1,2,3,4])
set_b = np.array([3,4,5,5])
intersection = np.intersect1d(set_a,set_b)
intersection
```

```python
union_of_array =np.union1d(set_a,set_b)
print("union of 1d array\n",union_of_array)
```

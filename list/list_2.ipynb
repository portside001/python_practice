# COPY List

```python
l = ['name',12345,"address",'test data']
```

```python
l1 = l 
```

```python
print(id(l1),'-->',id(l))
print(l1,'-->',l)
```

```python
l1[0] = 20
print(l1,'-->',l)
```

```python
# cretae shallow copy of list 
l1 = l.copy()
print(id(l1),'-->',id(l))
```

```python
print(l1,'-->',l)
```

```python
l1[0] = 1
print(l1,'-->',l)
```

```python
print(id(l1),'-->',id(l))
```

# All AND Any

- the all method returns:
- True -: if all elements in a list are true
- False -: if all elements in a list is false

```python
l = [1,2,3,4,5]
all(l)
```

```python
l=[False]
all(l)
```

- not even single value should be false

```python
l = [1,2,3,4,5,True,False]
all(l)
```

```python
any(l)
```

```python
l = [False]
```

```python
any(l)
```

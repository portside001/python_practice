# UNION 

- Union = "Sabko bulao, duplicate ko ek hi baar bulao." 😄

```python
s1 = {"jhon","cone","tone"}
s2 = {"tedy","mady","max","JHON","JHON"}

# WHEN WE WANT TO USE UNION WE NEED TO USE UNION METHOD OR BY USING PIPE OPERATOR | 


s1.union(s2)
```

```python
a = {1,2,3,4,5}
b = {4,5,6,7,8}
c = {8,9,10} 
```

```python
a.union(b)
```

```python
a.union(b,c)
```

```python
b.union(a, c)
```

```python
# for union we can use the | operator 
a | b | c
```

# Difference in Set (-)

- Take my elements and remove common elements.

```python
A = {1,2,3,4,5}
B = {4,5,6,7}
A - B 
```

- A = 1 2 3 4 5
- B =       4 5 6 7

- Remove common (4,5)
- Remaining → 1,2,3

```python
A
```

```python
B.difference(A)
```

# Symmetric Difference (^)

- Take ONLY unique elements from both sets.
- Remove common elements.

```python
A = {1,2,3,4,5}
B = {4,5,6,7}
A.symmetric_difference(B)
```

# Subset (issubset() or <=)

- "Everything in A is present in B."

```python
A ={1,2}
B = {1,2,3,4,5}
A.issubset(B)
```

# Superset (issuperset() or >=)

- "I contain all elements of another set."

```python
B = {1,2,3,4}
A = {1,2}

B.issuperset(A)
```

# Disjoint (isdisjoint())

```python
A = {1,2,3}
B = {4,5,6}

A.isdisjoint(B)
```

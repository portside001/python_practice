# Dictionary 

- Dictionary is mutable data type in Python 
- A Python Dictionary is a collection of key and value pair
- {key:value}
- key must be unique in a dictionary 
- duplicate values are allowed 

```python
# empty dictionary 
mydict = dict()
mydict 
```

```python
#  dict with integer keys
mydict = {1:'one',2:'two',3:'three'}
type(mydict)
```

```python
mydict = {1:1}
type(mydict)
```

```python
# dictionary with character keys
mydict = {'A':'one','B':'two','C':'three'}
mydict
```

```python
# dict with mixed keys 
mydict = {1:'one','A':'two',3:'three'}
mydict
```

```python
# return dictionary keys using keys() method 
mydict.keys()
```

```python
# return dictonary values using values method 
mydict.values()
```

```python
# Access each key-value pair with in a dictionary using items() method
mydict.items()
```

```python
# dictionary with list 
mydict = {1:'one',2:'two',3:['love','hate','drama']}
mydict
```

```python
# dictionary using list and tuple and set 
mydict = {1:'one',2:'two','A':{'love','hate','drama'},'B':['l1','l2','l3'],'C':('BAT','CAT','DAD')}
type(mydict)
```

```python
# create a dictionary from a sequence of keys
keys = {'a','b','c','d'}
print(type(keys))
mydict3 = dict.fromkeys(keys)
mydict3
```

```python
# create a dictionary from a sequence of key and value 
keys
value = 10
mydict4 = dict.fromkeys(keys,value)
mydict4
```

```python
keys
value = [10,20,30]
mydict4 = dict.fromkeys(keys,value)
mydict4
```

```python
# update a value 
value.append(40)
mydict4
```

# Important Interview Point 🚀

- Dictionary mutability depends on its values.

- Dictionary itself → Mutable ✅
- List inside dictionary → Mutable ✅
- Tuple inside dictionary → Immutable ❌

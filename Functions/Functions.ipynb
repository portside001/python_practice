```python
# find count odd and even
l = [10,9,8,3,5,6,1]
def count(l):
    even = 0
    odd = 0

    for i in l:
        if i%2 == 0: 
            even += 1
        else:
            odd += 1

    return even,odd
even,odd = count(l)    
print(f"total even number present in list {even}")
print(f"total odd number present in list {odd}")
```

```python
nums = [3,2,6,8,4,6,2,9]
def is_even(n):
    return n % 2 == 0
evens = list(filter(is_even, nums))

double = list(map(lambda n : n*2, evens))
double_ = list(map(lambda n : n+2, evens))
doubble_1 = list(map(lambda n : n-2, evens)) 

print(evens)
print(double)  
print(double_)
print(doubble_1)
```

```python
from functools import reduce

def add_all(a,b):
    return a+b

nums = [3,2,6,8,4,6,2]

evens = list(filter(is_even, nums))
double = list(map(lambda n : n*2, evens))

sums = reduce(add_all, double)
sums
print(sums)     
```

# filter -: SELECT SPECIFIED DATA

```python
num = [1,2,3,4,5,6]
even = list(filter(lambda x : x % 2 == 0,num))
print(even)
```

# map -:  Transorm the data 

```python
num = [1,2,3,4,5,6]
double = list(map(lambda x : x * 2,num))
print(double)
```

# Reduce -: to perform opertions like on group 
- for this we need to use functool to use reduce 

```python
from functools import reduce
```

```python
num1 = [1,2,3,4,5]
num2 = [5,4,3,2,1]

def add_all(x):
    return x % 2 == 0 

def sum(x,y):
    return x + y

evens_user_define = list(filter(add_all,num1))
evens_lambda = list(map(lambda x : x*2,num2))


sums = reduce(sum,num2)
sums_using_lambda = reduce(lambda x, y: x + y,num2)

print(evens_user_define)
print(evens_lambda)
print(sums)    
print(sums_using_lambda)
```

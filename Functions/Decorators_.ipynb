# DECORATORS 

- decorator is a higher order function which takes function as an arguments and returns an funtion 
- it perform some operations before and after without changing its original source code 

```python
def add(a,b):
    return a // b 

add(2,10)
```

- numerator should be greater then denomerator 

```python
def div(a,b):
    if a < b :
        a , b = b ,a
        return a // b
div(2,4)
```

- using decorator we can add extra feature in existing code

```python
# using help of the decorator you can add the extra feature in the exicting function 

def div(a,b):
    print(a / b)

def decorator_fun(func): # hear div_dectorator will accept the div function
    def wrapper(a,b):
        if a<b:
            a,b = b,a
        return func(a,b)
    return wrapper

div = decorator_fun(div)

div(2,4)     
```

```python
def my_decorator(func):

    def wrapper():
        print("Before the function")

        func()              # Call original function

        print("After the function")

    return wrapper

@my_decorator
def say_hello():
    print("Hello!")


say_hello()
```

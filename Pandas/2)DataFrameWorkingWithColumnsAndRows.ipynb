# 7. assign() Method

- df.assign(...)
- example 
- iris.assign(
    sepal_ratio=
    iris["SepalWidth"] /
    iris["SepalLength"]
)
- Very important for chaining operations.
- returns a copy.

- Original dataframe remains unchanged.

# 8. Using Lambda in assign()

```python
df
```

```python
df['name'] = df['name'].apply(lambda x: x.upper())
```

```python
df
```

```python
df['adult'] = df['age'].apply(lambda x: 'Yes' if x >= 18 else 'No')
```

```python
df
```

# 9. Multiple Column Creation

```python
df = df.assign(
    age_plus_10=lambda x: x['age'] + 10,
    adult=lambda x: x['age'] >= 18,
    name_upper=lambda x: x['name'].str.upper()
)
```

```python
df
```

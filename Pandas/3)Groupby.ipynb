# What is GroupBy?

```python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["Sachin","Rahul","Amit","Neha","Priya",
                 "Karan","Rohit","Pooja","Ankit","Simran"],

    "Department": ["IT","HR","IT","Finance","HR",
                   "Finance","IT","HR","Finance","IT"],

    "City": ["Indore","Delhi","Indore","Mumbai","Delhi",
             "Mumbai","Pune","Delhi","Mumbai","Pune"],

    "Salary": [50000,40000,60000,55000,45000,
               70000,65000,48000,72000,62000],

    "Experience": [2,1,4,3,2,6,5,2,7,4]
})

print(df)
```

```python
df
```

```python
df.groupby("Department").sum(numeric_only=True)
```

```python
df.groupby("Department")["Salary"].mean()
```

```python
df.groupby("Department")["Salary"].max()
```

```python
df.groupby("Department")['Employee'].count()
```

```python
df.groupby('Department')['Salary'].transform('mean')
```

```python
df
```

```python
df.groupby(level=0)
```

```python
df.groupby("Department").filter(
    lambda x: x["Salary"].mean() > 50000
)
```

```python
df
```

```python
df.groupby("Department")["Salary"].agg("mean")
```

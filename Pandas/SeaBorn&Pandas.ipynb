# 6 Box Plot?

A Box Plot (or Box-and-Whisker Plot) summarizes the distribution of a numerical variable using five important 
also known as 5 numbber summary values.

It answers questions like:

What is the median?
What is the spread of the data?
Are there outliers?
Is the data skewed?

```python
box_ploat_data_visulize = sns.boxplot(df,x="BirthRate") 
```

# 7 What is a Violin Plot?

A Violin Plot is used to visualize the distribution of a numerical variable and compare it across one or more categories.

It answers questions like:

What is the distribution of the data?
Where are most values concentrated?
What is the median?
Is the distribution symmetric or skewed?
Are there multiple peaks?

```python
sns.violinplot(df,x ='BirthRate')
```

```python
sns.violinplot(data=df, x='IncomeGroup', y='BirthRate')
plt.xticks(rotation=45)
```

# Strip Plot?

- A Strip Plot visualizes the distribution of a numerical variable across one or more categories by plotting each data point as a dot.

- It answers questions like:

- What are the actual values?
- Are there clusters?
- Are there outliers?
- How are values distributed within each category?

```python
sns.stripplot(df,x='BirthRate')
```

```python
sns.stripplot(data=df,
              x='IncomeGroup',
              y='BirthRate',
              jitter=True)
```

# Swarm Plot?

A Swarm Plot displays every individual data point of a numerical variable across one or more categories without overlapping points.

It answers questions like:

What are the actual observations?
How are they distributed?
Are there clusters?
Are there outliers?
How dense is each category?

```python
sns.swarmplot(df,x='BirthRate')
```

```python

sns.swarmplot(df,
              x='IncomeGroup',
              y='BirthRate',
              hue='IncomeGroup')
```

# Scatter Plot?

A Scatter Plot shows the relationship between two numerical variables.

Each dot represents one observation (one row) in the dataset.

It answers questions like:

Is there a relationship between two variables?
Does one variable increase when the other increases?
Is there any pattern?
Are there outliers?

```python
sns.scatterplot(data=df, x='BirthRate', y='IncomeGroup')
```

# Pair Plot?

A Pair Plot automatically creates a grid of plots to show:

Distribution of each numerical variable (on the diagonal)
Relationship between every pair of numerical variables (off-diagonal)

Instead of creating many scatter plots manually, one command creates them all.

```python
sns.pairplot(df)
```

# Heatmap?

A Heatmap is a color-coded matrix where colors represent values.

Instead of looking at numbers, you understand the data by looking at colors.

It answers questions like:

Which variables are highly correlated?
Which variables have weak correlation?
Which values are high or low?
Where are patterns in the data?

```python
corr = df.corr(numeric_only=True)
```

```python
corr
```

```python
sns.heatmap(corr)
```

```python
sns.heatmap(corr, annot=True)
```
